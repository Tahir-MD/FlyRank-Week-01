## 1. Distributions

*Look before deciding: distributions of your key fields. Note the heavy tails.*

```python
# This cell is for CODE (numbers, a query, a check).
import pandas as pd

df = pd.read_csv("/content/content_refresh_anonymized.csv")
print(df.shape)

key_cols = ["impressions_90d", "clicks_90d", "sessions_90d", "ctr",
            "avg_position", "engagement_rate", "trend_pct", "search_volume"]

print(df[key_cols].describe())

# heavy-tail check: compare median to the 99th percentile
for col in key_cols:
    print(col, "median:", df[col].median(), " p99:", df[col].quantile(0.99), " max:", df[col].max())
```

## 2. Signal test #1 / #2 / #3 (verdict each)

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

```python
# This cell is for CODE (numbers, a query, a check).
declining = df[df["trend_direction"] == "down"]
stable = df[df["trend_direction"] != "down"]

def compare(col):
    print(f"{col}: declining median = {declining[col].median():.3f} | "
          f"stable median = {stable[col].median():.3f}")

compare("engagement_rate")

compare("ctr")

compare("days_since_last_update")
```
