## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

These three numbers show real signal worth acting on: a meaningful share of pages are
declining right now, search volume alone tells us almost nothing about real traffic,
and position on the results page creates a large, measurable gap in clicks. Together
they support a review-prioritization approach rather than guesswork.

```python
import pandas as pd

df = pd.read_csv("/content/content_refresh_anonymized.csv")

declining_share = (df["trend_direction"] == "down").mean()
print(f"Share of pages currently declining: {declining_share:.2%}")

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")

visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean()
print(f"CTR gap between best and worst position tier: {ctr_by_pos.max() - ctr_by_pos.min():.4f}")
```
