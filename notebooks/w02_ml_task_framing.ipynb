## 1. My lane as an ML task (type)

My lane is best described as a scoring and ranking task. Instead of putting every page
into one fixed category, the goal is to give each page a risk score that reflects how
likely it is to be declining, and then rank all pages from highest risk to lowest. This
matches the actual outcome I want, which is an ordered list telling a content team which
pages to review first.

```python
# This cell is for CODE (numbers, a query, a check).
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining"] = (df["trend_direction"] == "down").astype(int)
print(df["is_declining"].value_counts())
```

## 5. Why ML beats a fixed rule here

A simple fixed rule, like flagging any page with low click through rate, misses how these
signals interact with each other. For example, a low click through rate might be normal
for a page sitting in a weak position tier, but concerning for a page sitting in a strong
position. Content type also changes what a normal click through rate even looks like, as
shown in week one. Writing a single if statement that accounts for all these overlapping
conditions at once quickly becomes unreadable and brittle, while a model can weigh several
signals together and rank pages more reliably.
